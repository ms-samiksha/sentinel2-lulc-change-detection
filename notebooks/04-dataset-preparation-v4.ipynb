```python
import os
import numpy as np
import pandas as pd
import rasterio

# ============================================================
# PATHS
# ============================================================

BASE = "/kaggle/input/datasets/mssamiksha30/sentinel2-lulc-change-detection/sentinel2_lulc_change_dataset_v3/sentinel2_lulc_change_dataset_v2"

LABEL_2020_DIR = os.path.join(BASE, "labels_2020")
LABEL_2025_DIR = os.path.join(BASE, "labels_2025")

METADATA_PATH = os.path.join(BASE, "metadata", "patch_metadata.csv")

OUT_BASE = "/kaggle/working/corrected_label_patches"
OUT_2020 = os.path.join(OUT_BASE, "labels_2020")
OUT_2025 = os.path.join(OUT_BASE, "labels_2025")

os.makedirs(OUT_2020, exist_ok=True)
os.makedirs(OUT_2025, exist_ok=True)

# ============================================================
# LOAD PATCH METADATA
# ============================================================

metadata = pd.read_csv(METADATA_PATH)

print("=" * 70)
print("PATCH METADATA")
print("=" * 70)
print("Rows:", len(metadata))
print("Columns:", list(metadata.columns))

assert len(metadata) == 1008
assert {"patch_id", "parent_tile", "split"}.issubset(metadata.columns)

# ============================================================
# EXISTING PATCH EXTRACTION RULE
# ============================================================
# 4 x 4 patches per tile
# Each patch = 128 x 128
# Regular stride = 128
#
# patch 0   1   2   3
# patch 4   5   6   7
# patch 8   9  10  11
# patch 12 13  14 15
# ============================================================

PATCH_SIZE = 128
PATCHES_PER_TILE = 16
GRID_SIZE = 4

# ============================================================
# CREATE LABEL PATCHES
# ============================================================

for _, row in metadata.iterrows():

    patch_id = str(row["patch_id"])
    parent_tile = str(row["parent_tile"])

    # Example:
    # tile_000_patch_00000
    # tile_000_patch_00001
    # ...
    local_patch = int(patch_id.split("_patch_")[-1]) % 16

    patch_row = local_patch // GRID_SIZE
    patch_col = local_patch % GRID_SIZE

    y = patch_row * PATCH_SIZE
    x = patch_col * PATCH_SIZE

    # ----------------------------
    # 2020
    # ----------------------------

    path20 = os.path.join(
        LABEL_2020_DIR,
        parent_tile + ".tif"
    )

    with rasterio.open(path20) as src:
        label20 = src.read(
            1,
            window=rasterio.windows.Window(
                x, y, PATCH_SIZE, PATCH_SIZE
            )
        )

    assert label20.shape == (128, 128)

    np.save(
        os.path.join(OUT_2020, patch_id + ".npy"),
        label20.astype(np.uint8)
    )

    # ----------------------------
    # 2025
    # ----------------------------

    path25 = os.path.join(
        LABEL_2025_DIR,
        parent_tile + ".tif"
    )

    with rasterio.open(path25) as src:
        label25 = src.read(
            1,
            window=rasterio.windows.Window(
                x, y, PATCH_SIZE, PATCH_SIZE
            )
        )

    assert label25.shape == (128, 128)

    np.save(
        os.path.join(OUT_2025, patch_id + ".npy"),
        label25.astype(np.uint8)
    )

# ============================================================
# VALIDATION
# ============================================================

files20 = sorted(
    f for f in os.listdir(OUT_2020)
    if f.endswith(".npy")
)

files25 = sorted(
    f for f in os.listdir(OUT_2025)
    if f.endswith(".npy")
)

print("\n" + "=" * 70)
print("CORRECTED LABEL PATCHES")
print("=" * 70)

print("2020:", len(files20))
print("2025:", len(files25))

assert len(files20) == 1008
assert len(files25) == 1008
assert set(files20) == set(files25)

metadata_ids = set(
    metadata["patch_id"].astype(str) + ".npy"
)

assert set(files20) == metadata_ids

print("✓ 1008 2020 patches")
print("✓ 1008 2025 patches")
print("✓ Patch IDs match metadata")

# ============================================================
# VALUE + SHAPE CHECK
# ============================================================

allowed = {0, 1, 2, 3, 255}

for folder, name in [
    (OUT_2020, "2020"),
    (OUT_2025, "2025")
]:

    for fname in files20:

        arr = np.load(
            os.path.join(folder, fname)
        )

        assert arr.shape == (128, 128)

        values = set(np.unique(arr).tolist())

        assert values.issubset(allowed), (
            f"Invalid values in {name}/{fname}: {values}"
        )

print("✓ All patches are 128 × 128")
print("✓ All label values are valid")

# ============================================================
# CLASS COUNTS
# ============================================================

for folder, name in [
    (OUT_2020, "2020"),
    (OUT_2025, "2025")
]:

    counts = {
        0: 0,
        1: 0,
        2: 0,
        3: 0,
        255: 0
    }

    for fname in os.listdir(folder):

        if not fname.endswith(".npy"):
            continue

        arr = np.load(
            os.path.join(folder, fname)
        )

        values, nums = np.unique(
            arr,
            return_counts=True
        )

        for v, n in zip(values, nums):
            counts[int(v)] += int(n)

    print(f"\n{name} class counts:")
    print("Water      (0)  :", counts[0])
    print("Vegetation (1)  :", counts[1])
    print("Built-up   (2)  :", counts[2])
    print("Bare Land  (3)  :", counts[3])
    print("NoData   (255)  :", counts[255])

print("\n" + "=" * 70)
print("✓ CORRECTED LABEL PATCH GENERATION COMPLETE")
print("=" * 70)

print("Output:", OUT_BASE)
```

```python
import os
import shutil
import numpy as np
import pandas as pd


# ============================================================
# PATHS
# ============================================================

# ------------------------------------------------------------
# CURRENT v3 DATASET
# Contains:
#   2020/
#   2025/
#   labels_2020/
#   labels_2025/
#   metadata/
# ------------------------------------------------------------
V3 = (
    "/kaggle/input/datasets/mssamiksha30/"
    "sentinel2-lulc-change-detection/"
    "sentinel2_lulc_change_dataset_v3/"
    "sentinel2_lulc_change_dataset_v2"
)


# ------------------------------------------------------------
# ORIGINAL v2 DATASET
# This is where the EXISTING IMAGE PATCHES are located:
#
# patches/
#   2020/
#   2025/
#   labels_2020/   <- OLD labels, DO NOT USE
# ------------------------------------------------------------
V2 = (
    "/kaggle/input/datasets/mssamiksha30/"
    "sentinel2-lulc-change-detection/"
    "sentinel2_lulc_change_dataset_v2"
)


# ------------------------------------------------------------
# CORRECTED LABEL PATCHES GENERATED IN THIS NOTEBOOK
#
# corrected_label_patches/
#   labels_2020/
#   labels_2025/
# ------------------------------------------------------------
CORRECTED = "/kaggle/working/corrected_label_patches"


# ------------------------------------------------------------
# STAGING FOLDER
# This is NOT uploaded to Kaggle yet.
# ------------------------------------------------------------
FINAL_CANDIDATE = (
    "/kaggle/working/"
    "sentinel2_lulc_change_dataset_v4_candidate"
)


# ============================================================
# CLEAN START
# ============================================================

if os.path.exists(FINAL_CANDIDATE):
    shutil.rmtree(FINAL_CANDIDATE)

os.makedirs(FINAL_CANDIDATE, exist_ok=True)


print("=" * 70)
print("BUILDING FINAL DATASET CANDIDATE FOR V4")
print("=" * 70)


# ============================================================
# 0. VERIFY INPUT SOURCES BEFORE COPYING
# ============================================================

print("\n[0/6] Checking source datasets...")


# Check v3
if not os.path.exists(V3):
    raise FileNotFoundError(
        f"v3 dataset not found:\n{V3}"
    )

print("✓ v3 dataset found")


# Check old v2 patches
V2_PATCHES = os.path.join(V2, "patches")

if not os.path.exists(V2_PATCHES):
    raise FileNotFoundError(
        f"Original v2 patches not found:\n{V2_PATCHES}"
    )

print("✓ original v2 patches found")


# Check corrected label patches
if not os.path.exists(CORRECTED):
    raise FileNotFoundError(
        f"Corrected label patches not found:\n{CORRECTED}"
    )

print("✓ corrected label patches found")


# Check required corrected label folders
for folder in ["labels_2020", "labels_2025"]:

    path = os.path.join(CORRECTED, folder)

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Missing corrected label folder:\n{path}"
        )

    count = len([
        f for f in os.listdir(path)
        if f.endswith(".npy")
    ])

    if count != 1008:
        raise ValueError(
            f"{folder}: expected 1008 patches, found {count}"
        )

    print(f"✓ corrected {folder}: {count} patches")


# ============================================================
# 1. COPY EVERYTHING FROM V3
# ============================================================

print("\n[1/6] Copying v3 dataset...")

shutil.copytree(
    V3,
    FINAL_CANDIDATE,
    dirs_exist_ok=True
)

print("✓ 2020 full tiles copied")
print("✓ 2025 full tiles copied")
print("✓ corrected 2020 full labels copied")
print("✓ corrected 2025 full labels copied")
print("✓ metadata copied")


# ============================================================
# 2. COPY EXISTING IMAGE PATCHES FROM ORIGINAL V2
# ============================================================

print("\n[2/6] Copying existing image patches from original v2...")


PATCHES_OUT = os.path.join(
    FINAL_CANDIDATE,
    "patches"
)

os.makedirs(
    PATCHES_OUT,
    exist_ok=True
)


for folder in ["2020", "2025"]:

    source = os.path.join(
        V2_PATCHES,
        folder
    )

    destination = os.path.join(
        PATCHES_OUT,
        folder
    )


    if not os.path.exists(source):
        raise FileNotFoundError(
            f"Could not find original image patches:\n{source}"
        )


    shutil.copytree(
        source,
        destination,
        dirs_exist_ok=True
    )


    count = len([
        f for f in os.listdir(destination)
        if f.endswith(".npy")
    ])


    if count != 1008:
        raise ValueError(
            f"patches/{folder}: expected 1008, found {count}"
        )


    print(
        f"✓ patches/{folder}: {count} files"
    )


# ============================================================
# 3. COPY CORRECTED LABEL PATCHES
# ============================================================

print("\n[3/6] Adding corrected label patches...")


for folder in ["labels_2020", "labels_2025"]:

    source = os.path.join(
        CORRECTED,
        folder
    )

    destination = os.path.join(
        PATCHES_OUT,
        folder
    )


    # If an old folder somehow exists, remove it.
    # This guarantees that ONLY corrected labels are used.
    if os.path.exists(destination):
        shutil.rmtree(destination)


    shutil.copytree(
        source,
        destination
    )


    count = len([
        f for f in os.listdir(destination)
        if f.endswith(".npy")
    ])


    if count != 1008:
        raise ValueError(
            f"patches/{folder}: expected 1008, found {count}"
        )


    print(
        f"✓ patches/{folder}: {count} corrected files"
    )


# ============================================================
# 4. GENERATE 16-CLASS TRANSITION + BINARY CHANGE LABELS
# ============================================================

print(
    "\n[4/6] Generating semantic transition "
    "and binary change labels..."
)


LABEL_2020_DIR = os.path.join(
    PATCHES_OUT,
    "labels_2020"
)

LABEL_2025_DIR = os.path.join(
    PATCHES_OUT,
    "labels_2025"
)

TRANSITION_DIR = os.path.join(
    PATCHES_OUT,
    "transition_labels"
)

CHANGE_DIR = os.path.join(
    PATCHES_OUT,
    "change_labels"
)


os.makedirs(
    TRANSITION_DIR,
    exist_ok=True
)

os.makedirs(
    CHANGE_DIR,
    exist_ok=True
)


files_2020 = sorted([
    f for f in os.listdir(LABEL_2020_DIR)
    if f.endswith(".npy")
])


files_2025 = sorted([
    f for f in os.listdir(LABEL_2025_DIR)
    if f.endswith(".npy")
])


# Make sure both years contain exactly the same patch IDs.
if files_2020 != files_2025:
    raise ValueError(
        "2020 and 2025 corrected label patch filenames do not match!"
    )


if len(files_2020) != 1008:
    raise ValueError(
        f"Expected 1008 label patches, found {len(files_2020)}"
    )


print(
    f"Generating targets for {len(files_2020)} patches..."
)


for i, filename in enumerate(files_2020):


    # --------------------------------------------------------
    # Load corrected labels
    # --------------------------------------------------------

    y2020 = np.load(
        os.path.join(
            LABEL_2020_DIR,
            filename
        )
    )

    y2025 = np.load(
        os.path.join(
            LABEL_2025_DIR,
            filename
        )
    )


    # --------------------------------------------------------
    # Shape validation
    # --------------------------------------------------------

    if y2020.shape != (128, 128):
        raise ValueError(
            f"{filename}: invalid 2020 shape "
            f"{y2020.shape}"
        )


    if y2025.shape != (128, 128):
        raise ValueError(
            f"{filename}: invalid 2025 shape "
            f"{y2025.shape}"
        )


    # --------------------------------------------------------
    # Valid pixels
    #
    # Valid classes:
    #   0 = Water
    #   1 = Vegetation
    #   2 = Built-up
    #   3 = Bare Land
    #
    # 255 = NoData / Ignore
    # --------------------------------------------------------

    valid = (
        (y2020 >= 0) & (y2020 <= 3) &
        (y2025 >= 0) & (y2025 <= 3)
    )


    # ========================================================
    # 16-CLASS SEMANTIC TRANSITION LABEL
    # ========================================================
    #
    # transition = 2020_class * 4 + 2025_class
    #
    # 0  = Water       -> Water
    # 1  = Water       -> Vegetation
    # 2  = Water       -> Built-up
    # 3  = Water       -> Bare Land
    #
    # 4  = Vegetation  -> Water
    # 5  = Vegetation  -> Vegetation
    # ...
    #
    # 15 = Bare Land   -> Bare Land
    #
    # Invalid pixels = 255
    # ========================================================

    transition = np.full(
        (128, 128),
        255,
        dtype=np.uint8
    )


    transition[valid] = (
        y2020[valid] * 4 +
        y2025[valid]
    ).astype(np.uint8)


    # ========================================================
    # 2-CLASS BINARY CHANGE LABEL
    # ========================================================
    #
    # 0   = No Change
    # 1   = Change
    # 255 = Ignore
    # ========================================================

    change = np.full(
        (128, 128),
        255,
        dtype=np.uint8
    )


    change[valid] = (
        y2020[valid] != y2025[valid]
    ).astype(np.uint8)


    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    np.save(
        os.path.join(
            TRANSITION_DIR,
            filename
        ),
        transition
    )


    np.save(
        os.path.join(
            CHANGE_DIR,
            filename
        ),
        change
    )


    if (i + 1) % 100 == 0:
        print(
            f"  Generated "
            f"{i + 1}/{len(files_2020)}"
        )


print(
    f"✓ {len(files_2020)} transition-label patches generated"
)

print(
    f"✓ {len(files_2020)} binary-change patches generated"
)


# ============================================================
# 5. COMPLETE PATCH VALIDATION
# ============================================================

print("\n[5/6] Running complete patch validation...")


EXPECTED = 1008


folders_to_check = [
    "2020",
    "2025",
    "labels_2020",
    "labels_2025",
    "transition_labels",
    "change_labels"
]


# ------------------------------------------------------------
# Check number of files
# ------------------------------------------------------------

for folder in folders_to_check:

    path = os.path.join(
        PATCHES_OUT,
        folder
    )


    files = sorted([
        f for f in os.listdir(path)
        if f.endswith(".npy")
    ])


    if len(files) != EXPECTED:
        raise ValueError(
            f"{folder}: expected "
            f"{EXPECTED}, found {len(files)}"
        )


    print(
        f"✓ patches/{folder}: {len(files)}"
    )


# ------------------------------------------------------------
# Check that every folder has exactly the same patch IDs
# ------------------------------------------------------------

file_sets = {}


for folder in folders_to_check:

    path = os.path.join(
        PATCHES_OUT,
        folder
    )


    file_sets[folder] = {
        f for f in os.listdir(path)
        if f.endswith(".npy")
    }


reference_set = file_sets["2020"]


for folder, files in file_sets.items():

    if files != reference_set:
        raise ValueError(
            f"Patch ID mismatch in {folder}"
        )


print(
    "✓ All six patch folders have "
    "identical patch IDs"
)


# ------------------------------------------------------------
# Validate arrays and label values
# ------------------------------------------------------------

transition_counts = np.zeros(
    16,
    dtype=np.int64
)

change_counts = np.zeros(
    2,
    dtype=np.int64
)


for filename in sorted(reference_set):


    y2020 = np.load(
        os.path.join(
            PATCHES_OUT,
            "labels_2020",
            filename
        )
    )


    y2025 = np.load(
        os.path.join(
            PATCHES_OUT,
            "labels_2025",
            filename
        )
    )


    transition = np.load(
        os.path.join(
            PATCHES_OUT,
            "transition_labels",
            filename
        )
    )


    change = np.load(
        os.path.join(
            PATCHES_OUT,
            "change_labels",
            filename
        )
    )


    # --------------------------------------------------------
    # Shape checks
    # --------------------------------------------------------

    for name, arr in [
        ("2020", y2020),
        ("2025", y2025),
        ("transition", transition),
        ("change", change)
    ]:

        if arr.shape != (128, 128):
            raise ValueError(
                f"{filename}: {name} has shape "
                f"{arr.shape}"
            )


    # --------------------------------------------------------
    # Allowed label values
    # --------------------------------------------------------

    if not np.isin(
        y2020,
        [0, 1, 2, 3, 255]
    ).all():

        raise ValueError(
            f"{filename}: invalid 2020 label value"
        )


    if not np.isin(
        y2025,
        [0, 1, 2, 3, 255]
    ).all():

        raise ValueError(
            f"{filename}: invalid 2025 label value"
        )


    if not np.isin(
        transition,
        list(range(16)) + [255]
    ).all():

        raise ValueError(
            f"{filename}: invalid transition value"
        )


    if not np.isin(
        change,
        [0, 1, 255]
    ).all():

        raise ValueError(
            f"{filename}: invalid change value"
        )


    # --------------------------------------------------------
    # Recalculate expected transition
    # --------------------------------------------------------

    valid = (
        (y2020 >= 0) & (y2020 <= 3) &
        (y2025 >= 0) & (y2025 <= 3)
    )


    expected_transition = (
        y2020[valid] * 4 +
        y2025[valid]
    ).astype(np.uint8)


    if not np.array_equal(
        transition[valid],
        expected_transition
    ):

        raise ValueError(
            f"{filename}: transition encoding mismatch"
        )


    # --------------------------------------------------------
    # Recalculate expected binary change
    # --------------------------------------------------------

    expected_change = (
        y2020[valid] != y2025[valid]
    ).astype(np.uint8)


    if not np.array_equal(
        change[valid],
        expected_change
    ):

        raise ValueError(
            f"{filename}: binary change mismatch"
        )


    # --------------------------------------------------------
    # NoData consistency
    # --------------------------------------------------------

    invalid = ~valid


    if not np.all(
        transition[invalid] == 255
    ):

        raise ValueError(
            f"{filename}: transition NoData mismatch"
        )


    if not np.all(
        change[invalid] == 255
    ):

        raise ValueError(
            f"{filename}: change NoData mismatch"
        )


    # --------------------------------------------------------
    # Counts
    # --------------------------------------------------------

    for c in range(16):

        transition_counts[c] += np.sum(
            transition == c
        )


    for c in range(2):

        change_counts[c] += np.sum(
            change == c
        )


# ============================================================
# 6. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("DATASET V4 CANDIDATE VALIDATION COMPLETE")
print("=" * 70)


print("\nPatch folders:")

for folder in folders_to_check:

    count = len([
        f
        for f in os.listdir(
            os.path.join(
                PATCHES_OUT,
                folder
            )
        )
        if f.endswith(".npy")
    ])

    print(
        f"  {folder:<20}: {count}"
    )


# ------------------------------------------------------------
# Transition distribution
# ------------------------------------------------------------

print(
    "\n16-class transition pixel counts:"
)

class_names = [
    "Water",
    "Vegetation",
    "Built-up",
    "Bare Land"
]


for c, count in enumerate(
    transition_counts
):

    from_class = c // 4
    to_class = c % 4

    print(
        f"  {c:2d}: "
        f"{class_names[from_class]:<11} → "
        f"{class_names[to_class]:<11}"
        f": {count:,}"
    )


# ------------------------------------------------------------
# Binary change distribution
# ------------------------------------------------------------

print(
    "\nBinary change pixel counts:"
)

print(
    f"  0 = No Change : "
    f"{change_counts[0]:,}"
)

print(
    f"  1 = Change    : "
    f"{change_counts[1]:,}"
)


# ------------------------------------------------------------
# Candidate location
# ------------------------------------------------------------

print(
    "\nCandidate location:"
)

print(FINAL_CANDIDATE)


print(
    "\n✓ ALL PATCH VALIDATIONS PASSED"
)

print(
    "✓ NO KAGGLE DATASET VERSION HAS BEEN CREATED"
)

print(
    "✓ V4 CANDIDATE IS READY FOR FINAL REVIEW"
)
```

```python
import os
import pandas as pd
import numpy as np

BASE = "/kaggle/working/sentinel2_lulc_change_dataset_v4_candidate"

print("=" * 70)
print("FINAL V4 DATASET STRUCTURAL CHECK")
print("=" * 70)

# ============================================================
# 1. FULL TILE COUNTS
# ============================================================

print("\n[1] Full-tile folders")

for folder in ["2020", "2025", "labels_2020", "labels_2025"]:

    path = os.path.join(BASE, folder)

    files = [
        f for f in os.listdir(path)
        if f.endswith(".tif")
    ]

    print(f"{folder:<15}: {len(files)}")

    if len(files) != 63:
        raise ValueError(
            f"{folder}: expected 63, found {len(files)}"
        )

print("✓ All four full-tile folders contain 63 TIFFs")


# ============================================================
# 2. PATCH COUNTS
# ============================================================

print("\n[2] Patch folders")

PATCHES = os.path.join(BASE, "patches")

patch_folders = [
    "2020",
    "2025",
    "labels_2020",
    "labels_2025",
    "transition_labels",
    "change_labels"
]

patch_sets = {}

for folder in patch_folders:

    path = os.path.join(PATCHES, folder)

    files = sorted([
        f for f in os.listdir(path)
        if f.endswith(".npy")
    ])

    print(f"{folder:<20}: {len(files)}")

    if len(files) != 1008:
        raise ValueError(
            f"{folder}: expected 1008, found {len(files)}"
        )

    patch_sets[folder] = set(files)

print("✓ All six patch folders contain 1008 files")


# ============================================================
# 3. PATCH ID CONSISTENCY
# ============================================================

print("\n[3] Patch ID consistency")

reference = patch_sets["2020"]

for folder, ids in patch_sets.items():

    if ids != reference:
        raise ValueError(
            f"Patch IDs do not match in {folder}"
        )

print("✓ All six patch folders contain identical patch IDs")


# ============================================================
# 4. METADATA
# ============================================================

print("\n[4] Patch metadata")

metadata_path = os.path.join(
    BASE,
    "metadata",
    "patch_metadata.csv"
)

if not os.path.exists(metadata_path):
    raise FileNotFoundError(
        f"Missing metadata:\n{metadata_path}"
    )

metadata = pd.read_csv(metadata_path)

print(f"Rows    : {len(metadata)}")
print(f"Columns : {list(metadata.columns)}")

if len(metadata) != 1008:
    raise ValueError(
        f"Expected 1008 metadata rows, found {len(metadata)}"
    )

required_columns = [
    "patch_id",
    "parent_tile",
    "split"
]

if list(metadata.columns) != required_columns:
    raise ValueError(
        f"Unexpected metadata columns: "
        f"{list(metadata.columns)}"
    )

metadata_ids = set(
    metadata["patch_id"].astype(str) + ".npy"
)

if metadata_ids != reference:
    raise ValueError(
        "Metadata patch IDs do not match patch filenames"
    )

print("✓ Metadata contains 1008 rows")
print("✓ Metadata columns are correct")
print("✓ Metadata patch IDs match the patches")


# ============================================================
# 5. SPLIT DISTRIBUTION
# ============================================================

print("\n[5] Dataset split")

print(
    metadata["split"]
    .value_counts()
    .to_string()
)

print("✓ Existing split metadata preserved")


# ============================================================
# 6. CHECK CORRECTED LABEL VALUES
# ============================================================

print("\n[6] Corrected label values")

for folder in ["labels_2020", "labels_2025"]:

    path = os.path.join(PATCHES, folder)

    values_found = set()

    for filename in os.listdir(path):

        if not filename.endswith(".npy"):
            continue

        arr = np.load(
            os.path.join(path, filename)
        )

        values_found.update(
            np.unique(arr).tolist()
        )

    allowed = {0, 1, 2, 3, 255}

    print(
        f"{folder:<15}: "
        f"{sorted(values_found)}"
    )

    if not values_found.issubset(allowed):
        raise ValueError(
            f"Invalid values found in {folder}"
        )

print("✓ Corrected label values are valid")


# ============================================================
# 7. CHECK TRANSITION VALUES
# ============================================================

print("\n[7] Transition labels")

transition_path = os.path.join(
    PATCHES,
    "transition_labels"
)

transition_values = set()

for filename in os.listdir(transition_path):

    if not filename.endswith(".npy"):
        continue

    arr = np.load(
        os.path.join(
            transition_path,
            filename
        )
    )

    transition_values.update(
        np.unique(arr).tolist()
    )

allowed_transition = set(range(16)) | {255}

print(
    "Values found:",
    sorted(transition_values)
)

if not transition_values.issubset(
    allowed_transition
):
    raise ValueError(
        "Invalid transition-label values found"
    )

print("✓ Transition labels contain only 0–15 and 255")


# ============================================================
# 8. CHECK BINARY CHANGE VALUES
# ============================================================

print("\n[8] Binary change labels")

change_path = os.path.join(
    PATCHES,
    "change_labels"
)

change_values = set()

for filename in os.listdir(change_path):

    if not filename.endswith(".npy"):
        continue

    arr = np.load(
        os.path.join(
            change_path,
            filename
        )
    )

    change_values.update(
        np.unique(arr).tolist()
    )

allowed_change = {0, 1, 255}

print(
    "Values found:",
    sorted(change_values)
)

if not change_values.issubset(
    allowed_change
):
    raise ValueError(
        "Invalid binary-change values found"
    )

print("✓ Binary change labels contain only 0, 1 and 255")


# ============================================================
# FINAL RESULT
# ============================================================

print("\n" + "=" * 70)
print("FINAL V4 STRUCTURAL CHECK PASSED")
print("=" * 70)

print("""
✓ 63 × 2020 full tiles
✓ 63 × 2025 full tiles
✓ 63 × corrected 2020 labels
✓ 63 × corrected 2025 labels

✓ 1008 × 2020 image patches
✓ 1008 × 2025 image patches
✓ 1008 × corrected 2020 label patches
✓ 1008 × corrected 2025 label patches
✓ 1008 × 16-class transition patches
✓ 1008 × binary change patches

✓ Patch IDs consistent
✓ Metadata consistent
✓ Train/validation/test split preserved
✓ Label values valid
✓ Transition values valid
✓ Binary-change values valid

READY TO PUBLISH V4
""")
```
